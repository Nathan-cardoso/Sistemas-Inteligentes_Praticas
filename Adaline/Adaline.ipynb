```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
ionosphere = fetch_ucirepo(id=52) 
  
# data (as pandas dataframes) 
X = ionosphere.data.features 
y = ionosphere.data.targets 
  
display(X)
display(y)
```

### Questão 1

```python
# Transformação dos dados numericos polares

y_bin = y['Class'].map({'g': 1, 'b': -1})

display(y_bin)

print("Distribuição das classes:")
print(y_bin.value_counts())

```

```python
# Verificação de valores ausentes
print(X.isnull().sum())
print(f"Total de ausentes: {X.isnull().sum().sum()}")

#Verificando possíveis valores constantes

std_to_columns = X.std()
const = std_to_columns[std_to_columns == 0].index.tolist()
print(f"\nAtributos constantes encontrados: {const}")

new_X = X.drop(columns=const)
print(f"Shape após remover constantes: {new_X.shape}")
```

```python
#Normalização Z-score
mean = new_X.mean()
desvio = new_X.std()

X_norm = (new_X - mean) / desvio

print(X_norm.mean().round(6))
print(X_norm.std().round(6))

display(X_norm)
display(y_bin)
```

## Justificativa a normalização:
Se os atributos tiverem escalas diferentes, isso irá impactar de forma negativa no gradiente descendente. Fazendo com que o treinamento tenha oscilações ou no momento da sua convergência, ela aconteça de forma lenta.

Assim, a normalização é muito importante para tornar todos os atributos na mesma escala e evitar as oscilições e erros na convergência.

### Questão 2

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_norm, y_bin,
    test_size=0.2,
    stratify=y_bin,
    random_state=42
)

etas = [1e-4, 1e-3, 1e-2, 5e-2, 1e-1]

for eta in etas:
    modelo = SGDRegressor(
        loss="squared_error",
        learning_rate="constant",
        eta0=eta,
        max_iter=1,
        tol=None,
        random_state=42
    )

    losses = []

    for epoch in range(200):
        modelo.partial_fit(X_train, y_train)

        y_pred = modelo.predict(X_train)

        # Função de perda do Adaline (MSE / 2)
        loss = ((y_train - y_pred) ** 2).mean() / 2
        losses.append(loss)

    plt.plot(losses, label=f"η = {eta}")
       
plt.xlabel("Épocas")
plt.ylabel("Função de perda")
plt.title("Curva de convergência - Adaline")
plt.legend()
plt.grid()

plt.show()
```

### Questão 3

```python
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

best_eta = 1e-2

model = SGDRegressor(
    loss="squared_error",
    learning_rate="constant",
    eta0=best_eta,
    max_iter=200,
    tol=None,
    random_state=42
)

model.fit(X_train, y_train)

y_pred_continuo = model.predict(X_test)

# Converter saída contínua classe (+1 / -1)
y_pred = [1 if y >= 0 else -1 for y in y_pred_continuo]


matriz   = confusion_matrix(y_test, y_pred)
acuracia = accuracy_score(y_test, y_pred)
precisao = precision_score(y_test, y_pred, pos_label=1)
recall   = recall_score(y_test, y_pred, pos_label=1)
f1       = f1_score(y_test, y_pred, pos_label=1)

# Resultados
print("Matriz de Confusão:")
print(matriz)
print("\nMétricas:")
print(f"Acurácia:  {acuracia:.4f}")
print(f"Precisão:  {precisao:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")
```
